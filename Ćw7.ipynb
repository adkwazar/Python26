```python
class Osoba:
    def __init__(self, imie, wiek):
        self.imie = imie
        self.wiek = wiek
    
    def przedstaw_sie(self):
        return f'Witaj! Nazywam sie {self.imie}.'
    
    def ile_mam_lat(self):
        return f'Mam {self.wiek} lat.'
```

<h4> >> Dziedziczenie

- klasa przejmuje atrybuty i metody innej klasy (nie trzeba pisać tego samego kodu ponownie)
- idea: klasa rodzic jest bardziej ogólna, klasy dzieci uszczegóławiają klasę rodzic; w ogólności można też dziedziczyć po wielu klasach

```python
class Student(Osoba): #klasa Student dziedziczy po klasie Osoba
    def __init__(self, imie, wiek, rok):
        super().__init__(imie, wiek) #pola imie i wiek jak w klasie Osoba (wywołuje init z klasy Osoba)
        self.rok = rok
    def ktory_rok(self):
        return f'Jestem na {self.rok} roku studiów.'
```

<h4> >> Metody korzystające z innych metod

```python
class Dzialania:
    def __init__(self, a, b):
        self.a = a
        self.b = b
    
    def dodawanie(self):
        return self.a+self.b
    
    def odejmowanie(self):
        return self.a-self.b
    
    def roznica_kwadratow(self):
        return self.dodawanie()*self.odejmowanie() #wykorzystuje wynik metod dodawania i odejmowania; a^2-b^2 = (a-b)(a+b)
    

x = Dzialania(5, 3)
print(x.dodawanie())
print(x.odejmowanie())
print(x.roznica_kwadratow())
```

<h4> >> Metody specjalne

Metody specjalne to wbudowane metody klas, których nazwy zaczynają i kończą się podwójnym podkreśleniem (np. `__init__`). Pozwalają definiować, jak obiekty zachowują się podczas wykonywania standardowych operacji w Pythonie.

### Najważniejsze metody specjalne:

| Metoda | Zastosowanie |
|---|---|
| `__init__(self)` | Inicjalizuje obiekt po jego utworzeniu (ustawia atrybuty). |
| `__str__(self)` | Określa tekstową reprezentację obiektu dla `print()`. |

| `__len__(self)` | Definiuje zachowanie funkcji `len()`. |
| `__add__(self, other)` | Definiuje zachowanie operatora `+` dla obiektów. |
| `__eq__(self, other)` | Definiuje porównywanie obiektów operatorem `==`. |
| `__getitem__(self, key)` | Umożliwia pobieranie elementów przez indeksowanie `[]`.

- Nie wywołuje się ich zazwyczaj bezpośrednio, lecz poprzez operacje i funkcje, np. `+`, `len()`, `print()`.

<h4> Przykład: Klasa Punkt

```python
#Klasa obsługująca punkty

class Punkt:
    def __init__(self, x=0, y=0):
        self.x = x
        self.y = y

    def __str__(self):  
        return f'pierwsza wsp.: {self.x}, druga wsp.: {self.y}'

    def __add__(self, other):    
        return Punkt(self.x + other.x, self.y + other.y) #TWORZY OBIEKT TEJ KLASY!!!
    
    def __eq__(self, other):  
        return self.x == other.x and self.y == other.y
```

```python
p1 = Punkt(4,5)
p2 = Punkt(1,2)
```

* ***METODA STATYCZNA*** - działa jak zwykła funkcja w orębie klasy, nie korzysta bezpośrednio z obiektu

```python
class DNA:
    @staticmethod 
    def komplementarna(sekwencja):
        kod = {"A": "T", "T": "A", "C": "G", "G": "C"}
        return "".join(kod[zasada] for zasada in sekwencja)

#Tutaj zasadniczo nie potrzebuje @staticmethod, zadziała i tak

dna = "ATGCCGTA"
print(DNA.komplementarna(dna)) 
```

```python
class DNA:
    @staticmethod 
    def komplementarna(sekwencja):
        kod = {"A": "T", "T": "A", "C": "G", "G": "C"}
        return "".join(kod[zasada] for zasada in sekwencja)


#MOGĘ TEŻ TAK, w te wersji bez @staticmethod nie zadziała  (a tak najczęsciej korzystamy pozniej z klas)

dna = DNA() #tworze obiekt klasy DNA
print(dna.komplementarna("ATGCCGTA")) 
```

* ***METODA KLASOWA*** - otrzymuje jako argument cls (całą klasę), może korzystać z atrybutów klasowych; działa na poziomie klasy

```python
#PRZYKŁAD 1

class DNA:
    nt = {"A", "T", "G","C"}
    
    @classmethod
    def czyok(cls, seq): 
        return set(seq).issubset(cls.nt)


dna = DNA()
print(dna.czyok("ATGCGAZ"))
```

```python
#PRZYKŁAD 2

class DNA:
    gatunek = "czlowiek"

    @classmethod
    def zmien_gatunek(cls, nowy_gatunek): #cls pracuje na całej klasie, self pracuje na konkretnym obiekcie
        cls.gatunek = nowy_gatunek


DNA.zmien_gatunek("szympans")

print(DNA.gatunek)  # szympans
```

* Dekorator ***@property*** - pozwala wywoływać metodę jak atrybut

```python
class DNA:
    def __init__(self, sekwencja):
        self.sekwencja = sekwencja

    
    def dlugosc(self):
        return len(self.sekwencja)

    @property
    def dlugosc2(self):
        return len(self.sekwencja)

dna = DNA("ATGC")
print(dna.dlugosc()) #wywoluje jak metode
print()
print(dna.dlugosc2)  #wywoluje jak atrybut
```
