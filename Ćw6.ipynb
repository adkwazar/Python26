<h4> Grupy

pozwalają wydzielić fragmenty dopasowanego tekstu, a następnie odwoływać się do nich lub przetwarzać je niezależnie

````python
#Zadanie: zwielokrotnij wszystkie wystapienia TA, ktore leżą pomiędzy A i G i zamień kolejnością G i A

import re
tekst = "GATAGCCTTT"   
tekst2 = re.sub(r'(A)(TA)(G)', lambda x: x.group(3)+x.group(2)*4+x.group(1), tekst)   #()()() #to co w pierwszym () to group(1) itd
print(tekst2)
````

````python
#Zadanie: wyszkuj obok siebie ten sam tekst występujący dwa razy po sobie

text = "11 1122 1134 abab 123123"
match = re.findall(r"(.+)\1", text)  #\1 znaczy to samo co wyzej group(1); (.+)\1 oznacza pewien tekst (.+), a następnie powtórzenie tego tekstu
print(match)
````

````python
text = "11 1122 1134 abab 123123" 
match = re.findall(r"(.+) \1", text) #tuta szukamy: tekst _(spacja)_ ten sam tekst
print(match)
````

<h4> >> Nakładające się powtórzenia

````python
seq = "ATATAT"
````

````python
re.findall(r'ATA', seq)
````

````python
#jak zlokalizować? Przypomnienie z poprzednich ćwiczeń
[(match.span(),match.group()) for match in re.finditer(r'ATA', seq)]
````

````python
re.findall(r"(?=(ATA))", seq) #dzięki (?=...) dopasowanie nie przesuwa pozycji wyszukiwania, dlatego możliwe jest znalezienie nakładających się motywów; nie myl z innym zastowaniem ? np w kontekście X? (znak występuje lub nie), umieszczenie ATA w grupie, tzn (ATA) pozwala np pozniej odczytac gdzie znajduje sie dany motyw
````

````python
#jak zlokalizować? Odnosimy się do pierszej grupy
[(match.span(1),match.group(1)) for match in re.finditer(r'(?=(ATA))', seq)]
````

<h3> Flagi w re

Flagi pozwalają zmieniać sposób działania wyrażeń regularnych, np. ignorować wielkość liter (re.IGNORECASE), dopasowywać wzorce w wielu liniach (re.MULTILINE) czy traktować tekst jako wieloliniowy (re.DOTALL).

<h4> re.IGNORCASE

````python
x = "ATGCActacacTA"
````

````python
r1 = re.findall(r"[tc]a", x) 
print(r1)
````

````python
r1 = re.findall(r"[tc]a", x, re.I) #re.I = re.IGNORCASE - ignoruj wielkość liter
print(r1)
````

<h4> re.MULTILINE

````python
x = """abc
def
adm
gam
efg
abc
ghi"""
````

````python
print(x)
````

````python
r2 = re.findall(r"a[a-z]+", x) 
print(r2)
````

````python
r3 = re.findall(r"^a[a-z]+", x)
print(r3)
````

````python
r4 = re.findall(r"^a[a-z]+", x, re.M) #re.M  = re.MULTILINE, sprawdzi czy zaczyna się ^ lub kończy $ w każdej linijce
print(r4)
````

<h4> re.DOTALL

````python
x = """once upon a time,
there lived a king"""
````

````python
print(x)
````

````python
r5 = re.findall(r".+", x) 
print(r5)
````

````python
r6 = re.findall(r".+", x, re.S)  #re.S =re.DOTALL , wtedy  . objejmuje tez nową linię
print(r6)
````
